# Exercise 8: Train and Evaluate Word2Vec (CBOW) Embeddings for Semantic Similarity Analysis

## Aim

To implement Word2Vec CBOW embeddings and evaluate semantic similarity using learned word vectors.

## Objective

Understand CBOW, train a model, generate word vectors, retrieve similar words and interpret similarity scores.

## Pre-Lab Task:

### 1. What is Word2Vec?

Word2Vec learns dense vector representations based on word-context patterns.

### 2. What is CBOW?

Continuous Bag of Words predicts a target word from its surrounding context.

### 3. What is the meaning of `sg=0` in Gensim?

In Gensim's Word2Vec implementation, `sg=0` selects CBOW and `sg=1` selects Skip-Gram.

### 4. What is semantic similarity?

It estimates how related two words are in meaning or usage.

### 5. Why can small corpora be unreliable?

A small corpus provides few context examples, so learned similarities may not generalize.

## In- Lab Task

### Tasks

1. Prepare and tokenize a corpus.
2. Train Word2Vec using CBOW.
3. Display vocabulary and word embeddings.
4. Find words similar to a selected term.
5. Calculate similarity for word pairs.

#### 1. Create and tokenize an astronomy corpus.

```python
import re
from gensim.models import Word2Vec
corpus = [
    "bread is baked in an oven", "flour and yeast make bread",
    "chefs prepare soup and salad", "vegetables cook in a pan",
    "bread and soup are common foods", "bakers use flour for cakes"
]
tokens = [re.findall(r"\\b\\w+\\b", s.lower()) for s in corpus]
for row in tokens:
    print(row)
```

#### 2. Train the CBOW model.

```python
cbow = Word2Vec(sentences=tokens, vector_size=48, window=3, min_count=1,
                workers=1, sg=0, seed=24, epochs=120)
print("Model trained.")
print("Vocabulary:", sorted(cbow.wv.key_to_index))
```

#### 3. Display word embeddings.

```python
for word in ["planet", "star"]:
    print(word, "vector:", cbow.wv[word])
    print("Dimension:", len(cbow.wv[word]))
```

#### 4. Find words similar to planet.

```python
print(cbow.wv.most_similar("planet", topn=5))
```

#### 5. Calculate similarity between word pairs.

```python
for first, second in [("planet","orbit"), ("star","galaxy")]:
    print(f"Similarity({first}, {second}) = {cbow.wv.similarity(first, second):.4f}")
```

## Post Lab

### Live Scenario: Training CBOW on a cooking corpus

Replace the astronomy topic with cooking and inspect the words the trained CBOW model places near “bread”.

#### Sample Dataset

```python
cooking = [
    "bread is baked in an oven", "flour and yeast make bread",
    "chefs prepare soup and salad", "vegetables cook in a pan",
    "bread and soup are common foods", "bakers use flour for cakes"
]
cook_tokens = [re.findall(r"\\b\\w+\\b", s.lower()) for s in cooking]
cook_model = Word2Vec(cook_tokens, vector_size=32, window=2, min_count=1,
                      workers=1, sg=0, seed=8, epochs=100)
print("Vocabulary:", sorted(cook_model.wv.key_to_index))
print("Words similar to bread:", cook_model.wv.most_similar("bread", topn=3))
```
