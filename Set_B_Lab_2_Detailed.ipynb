## Aim

To perform NLP preprocessing techniques including tokenization, stop-word removal, stemming, lemmatization, POS tagging and Named Entity Recognition.

## Objective

Understand and apply common preprocessing methods to a sentence corpus.

## Pre-Lab Task:

### 1. What is tokenization?

Tokenization divides text into smaller units such as sentences or words.

### 2. What are stop words?

Stop words are frequent words such as “the”, “is” and “and” that may be removed when they do not help a particular analysis.

### 3. What is stemming?

Stemming removes word endings using simple rules. Its result may not always be a dictionary word.

### 4. What is lemmatization?

Lemmatization converts a word to a dictionary form using linguistic information.

### 5. What is POS tagging?

Part-of-speech tagging assigns labels such as noun, verb, adjective or adverb to tokens.

### 6. What is NER?

Named Entity Recognition detects and labels entities such as people, organizations, locations and dates.

## In- Lab Task

### Tasks

1. Create a sentence corpus.
2. Tokenize sentences and words.
3. Remove stop words.
4. Apply stemming and lemmatization.
5. Display POS tags.
6. Run named entity recognition.

#### 1. Import NLTK and download required resources.

```python
sentences = [
    "Leela visited the Science Museum in Bengaluru.",
    "The guide explained ancient navigation tools.",
    "Students wrote a report about the exhibition."
]
print("Dataset:", sentences)
```

#### 2. Create a new sentence corpus.

```python
sentences = [
    "Mira attended a marine biology workshop in Kochi.",
    "Researchers studied coral reefs near the island.",
    "Students prepared notes about ocean pollution."
]
print(sentences)
```

#### 3. Tokenize the text.

```python
from nltk.tokenize import word_tokenize, sent_tokenize
for sentence in sentences:
    print("Sentence tokens:", word_tokenize(sentence))
```

#### 4. Remove stop words and punctuation.

```python
from nltk.corpus import stopwords
stop_set = set(stopwords.words("english"))
words = word_tokenize(" ".join(sentences))
filtered = [w for w in words if w.isalpha() and w.lower() not in stop_set]
print(filtered)
```

#### 5. Apply stemming and lemmatization.

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
print("Stems:", [stemmer.stem(w.lower()) for w in filtered])
print("Lemmas:", [lemmatizer.lemmatize(w.lower()) for w in filtered])
```

#### 6. Perform POS tagging.

```python
from nltk import pos_tag
print(pos_tag(filtered))
```

#### 7. Perform Named Entity Recognition.

```python
from nltk import ne_chunk
tagged = pos_tag(word_tokenize(sentences[0]))
print(ne_chunk(tagged))
```

## Post Lab

### Live Scenario: Preprocessing an environmental report

A college environmental club is preparing a short report about coastal pollution. Apply preprocessing to the new sample and compare the token list with the filtered words.

#### Sample Dataset

```python
post_text = "Asha visited the Coastal Research Centre in Chennai. The scientists studied plastic waste."
post_tokens = word_tokenize(post_text)
post_filtered = [w for w in post_tokens if w.isalpha() and w.lower() not in stop_set]
print("Original:", post_text)
print("Tokens:", post_tokens)
print("Filtered:", post_filtered)
print("Stems:", [stemmer.stem(w.lower()) for w in post_filtered])
print("Lemmas:", [lemmatizer.lemmatize(w.lower()) for w in post_filtered])
print("POS tags:", pos_tag(post_filtered))
print("NER:", ne_chunk(pos_tag(word_tokenize(post_text))))
```
