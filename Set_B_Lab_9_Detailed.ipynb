# Exercise 9: Build an LSTM Model for Sentiment Classification

## Aim

To build and evaluate an LSTM model for classifying text sentiment into Positive and Negative classes.

## Objective

Tokenize and pad text, create numeric sequences, build a Keras LSTM model, train it and predict sentiment.

## Pre-Lab Task:

### 1. What is sentiment analysis?

Sentiment analysis identifies the opinion or polarity expressed in a text.

### 2. What is an RNN?

A Recurrent Neural Network processes sequences while carrying information across time steps.

### 3. What is an LSTM?

Long Short-Term Memory is an RNN architecture with gates designed to manage information over longer sequences.

### 4. Why tokenize and pad text?

Tokenization converts words to IDs, while padding makes sequences a common length for batch training.

### 5. What does a sigmoid output represent in binary classification?

It returns a value between zero and one that can be interpreted as the estimated probability of the positive class.

## In- Lab Task

### Tasks

1. Create labelled review examples.
2. Tokenize and pad reviews.
3. Build an Embedding-LSTM-Dense model.
4. Train and evaluate the model.
5. Predict the sentiment of a new review.

#### 1. Import TensorFlow and prepare a new app-review dataset.

```python
texts = [
    "Lessons are clear and helpful", "The portal loads very slowly",
    "Videos play without problems", "Login fails every time",
    "The notes are well organized", "The page crashes often",
    "Practice quizzes are useful", "The site freezes on startup"
]
labels = np.array([1,0,1,0,1,0,1,0], dtype=np.float32)
print("Dataset loaded. Number of examples:", len(texts))
```

#### 2. Convert text to sequences and pad them.

```python
tokenizer = Tokenizer(oov_token="<OOV>")
tokenizer.fit_on_texts(texts)
sequences = tokenizer.texts_to_sequences(texts)
max_len = max(len(seq) for seq in sequences)
X = pad_sequences(sequences, maxlen=max_len, padding="post")
print("Vocabulary size:", len(tokenizer.word_index) + 1)
print("Padded matrix shape:", X.shape)
print(X)
```

#### 3. Build the LSTM model.

```python
model = Sequential([
    Embedding(input_dim=len(tokenizer.word_index)+1, output_dim=16),
    LSTM(16),
    Dropout(0.2),
    Dense(1, activation="sigmoid")
])
model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
model.summary()
```

#### 4. Train and evaluate the model.

```python
history = model.fit(X, labels, epochs=25, batch_size=2, verbose=0)
loss, accuracy = model.evaluate(X, labels, verbose=0)
print("Training loss:", round(float(loss), 4))
print("Training accuracy:", round(float(accuracy), 4))
print("This small training score is not a reliable generalization estimate.")
```

#### 5. Predict sentiment for an unseen review.

```python
sample = "The application works perfectly"
sample_seq = tokenizer.texts_to_sequences([sample])
sample_pad = pad_sequences(sample_seq, maxlen=max_len, padding="post")
probability = float(model.predict(sample_pad, verbose=0)[0][0])
print("Review:", sample)
print("Positive probability:", round(probability, 4))
print("Predicted sentiment:", "Positive" if probability >= 0.5 else "Negative")
```

## Post Lab

### Live Scenario: Classifying feedback about an online learning portal

A college wants to estimate whether short portal reviews are positive or negative. Use a second small labelled dataset to demonstrate the workflow.

#### Sample Dataset

```python
portal_reviews = [
    "Lessons are clear and helpful", "The portal loads very slowly",
    "Videos play without problems", "Login fails every time",
    "The notes are well organized", "The page crashes often"
]
portal_labels = np.array([1,0,1,0,1,0], dtype=np.float32)
portal_tokenizer = Tokenizer(oov_token="<OOV>")
portal_tokenizer.fit_on_texts(portal_reviews)
portal_seq = portal_tokenizer.texts_to_sequences(portal_reviews)
portal_len = max(len(x) for x in portal_seq)
portal_X = pad_sequences(portal_seq, maxlen=portal_len, padding="post")
portal_model = Sequential([
    Embedding(input_dim=len(portal_tokenizer.word_index)+1, output_dim=12),
    LSTM(12), Dense(1, activation="sigmoid")
])
portal_model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
portal_model.fit(portal_X, portal_labels, epochs=20, batch_size=2, verbose=0)
new_review = "The lessons are useful"
new_X = pad_sequences(portal_tokenizer.texts_to_sequences([new_review]), maxlen=portal_len, padding="post")
p = float(portal_model.predict(new_X, verbose=0)[0][0])
print(new_review, "->", "Positive" if p >= 0.5 else "Negative", "(score:", round(p,3), ")")
```
