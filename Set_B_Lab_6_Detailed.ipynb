# Exercise 6: Word Embeddings using Word2Vec

## Aim

To understand Word2Vec and generate word embeddings for measuring semantic similarity between words and sentences.

## Objective

Understand embeddings, compare CBOW and Skip-Gram, train a model, retrieve vectors and calculate sentence similarity.

## Pre-Lab Task:

### 1. What is a word embedding?

A word embedding is a dense numerical vector learned to represent a word based on patterns in its usage.

### 2. What is Word2Vec?

Word2Vec is a family of methods that learns word vectors from surrounding context.

### 3. What does CBOW predict?

CBOW predicts a target word from its surrounding context words.

### 4. What does Skip-Gram predict?

Skip-Gram predicts context words from a target word.

### 5. What is cosine similarity?

Cosine similarity measures the angle-based similarity between two vectors.

## In- Lab Task

### Tasks

1. Create and tokenize a new corpus.
2. Train Word2Vec with Skip-Gram.
3. Display vocabulary and a sample vector.
4. Find similar words.
5. Average vectors to represent sentences.
6. Calculate cosine similarity.

#### 1. Import libraries and prepare a weather corpus.

```python
import re
import numpy as np
from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity

sentences = [
    "guitar players practice music", "piano melodies sound gentle",
    "musicians perform songs on stage", "guitar strings produce notes",
    "the piano plays a melody", "music concerts attract audiences",
    "doctors care for patients", "nurses work in hospitals"
]
tokens = [re.findall(r"\\b\\w+\\b", s.lower()) for s in sentences]
for row in tokens:
    print(row)
```

#### 2. Train a Skip-Gram Word2Vec model.

```python
model = Word2Vec(sentences=tokens, vector_size=40, window=3, min_count=1,
                  workers=1, sg=1, seed=31, epochs=100)
print("Vocabulary:", sorted(model.wv.key_to_index))
```

#### 3. Display an embedding vector.

```python
print("Vector for rain:")
print(model.wv["rain"])
print("Vector size:", len(model.wv["rain"]))
```

#### 4. Find words similar to rain.

```python
print(model.wv.most_similar("rain", topn=5))
```

#### 5. Create sentence vectors and calculate cosine similarity.

```python
def sentence_vector(sentence):
    words = re.findall(r"\\b\\w+\\b", sentence.lower())
    vectors = [model.wv[w] for w in words if w in model.wv]
    return np.mean(vectors, axis=0) if vectors else np.zeros(model.vector_size)

s1 = "rain falls from clouds"
s2 = "storms bring winds"
v1, v2 = sentence_vector(s1), sentence_vector(s2)
score = cosine_similarity(v1.reshape(1,-1), v2.reshape(1,-1))[0,0]
print("Sentence 1:", s1)
print("Sentence 2:", s2)
print("Cosine similarity:", round(float(score), 4))
```

## Post Lab

### Live Scenario: Exploring words from a gardening guide

A gardening guide uses related words in similar contexts. Train a second model on a new small corpus and inspect words similar to “garden”.

#### Sample Dataset

```python
garden_sentences = [
    "garden flowers need sunlight", "garden soil supports plants",
    "plants grow in healthy soil", "flowers bloom in spring",
    "gardeners water flowers daily", "sunlight helps plants grow"
]
garden_tokens = [re.findall(r"\\b\\w+\\b", s.lower()) for s in garden_sentences]
garden_model = Word2Vec(garden_tokens, vector_size=32, window=2, min_count=1,
                        workers=1, sg=1, seed=8, epochs=100)
print("Vocabulary:", sorted(garden_model.wv.key_to_index))
print("Words similar to garden:", garden_model.wv.most_similar("garden", topn=4))
```
