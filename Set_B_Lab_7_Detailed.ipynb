## Pre-Lab Task:

### 1. What is GloVe?

GloVe stands for Global Vectors for Word Representation and learns vectors from global word co-occurrence statistics.

### 2. What is a co-occurrence matrix?

It records how often words appear near one another within a corpus.

### 3. How is GloVe different from Word2Vec?

GloVe uses global co-occurrence information, while Word2Vec learns by predicting words from local context.

### 4. Why use pretrained embeddings?

They allow applications to reuse vectors learned from a large corpus instead of training from scratch.

### 5. Name two applications of embeddings.

Examples include semantic search, document similarity and text classification.

#### 1. Import libraries and set the vector-file path.

```python
s1 = "students read books in a library"
s2 = "learners borrow novels from a reading room"
print("Sentence pair ready.")
```

#### 2. Load the GloVe word vectors.

```python
def load_glove(path):
    vectors = {}
    with open(path, encoding="utf-8") as handle:
        for line in handle:
            parts = line.rstrip().split()
            if len(parts) > 2:
                try:
                    vectors[parts[0]] = np.asarray(parts[1:], dtype=np.float32)
                except ValueError:
                    continue
    return vectors

glove = load_glove(GLOVE_FILE) if os.path.isfile(GLOVE_FILE) else {}
print("Loaded vectors:", len(glove))
if not glove:
    print("Download a GloVe text file and update GLOVE_FILE before continuing.")
```

#### 3. Define a sentence embedding function.

```python
def sentence_embedding(sentence):
    found = []
    for word in sentence.lower().split():
        word = word.strip(".,!?;:")
        if word in glove:
            found.append(glove[word])
    return np.mean(found, axis=0) if found else None
print("Sentence embedding function defined.")
```

#### 4. Compare two music-themed sentences.

```python
s1 = "the pianist played a beautiful melody"
s2 = "the musician performed a lovely tune"
if glove:
    a, b = sentence_embedding(s1), sentence_embedding(s2)
    if a is not None and b is not None:
        print("Sentence 1:", s1)
        print("Sentence 2:", s2)
        print("Cosine similarity:", round(float(cosine_similarity(a.reshape(1,-1), b.reshape(1,-1))[0,0]), 4))
    else:
        print("At least one sentence has no words found in the loaded vectors.")
else:
    print("Similarity is skipped until the GloVe file is loaded.")
```
