## Pre-Lab Task:

### 1. What is BERT?

BERT is a transformer-based language model pretrained to learn contextual representations from text.

### 2. What is Named Entity Recognition?

NER identifies text spans and assigns entity categories such as person, organization or location.

### 3. What is fine-tuning?

Fine-tuning continues training a pretrained model on task-specific labelled examples.

### 4. Why use a pretrained model?

Pretraining provides representations learned from a large text collection and reduces the amount of task-specific training needed.

### 5. Does running a pretrained pipeline mean custom fine-tuning was performed?

No. Loading a model and predicting entities is inference. Custom fine-tuning requires labelled training data and a training process.
